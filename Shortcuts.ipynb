## This notebook explains the shortcuts used in jupyter lab notebook

#### Shortcuts to create markdown cells (cells which are highlighted in blue font texts). 
These markdown cells are like headings,they do not execute any code.

Advantage of these cells are you don't need to use # to write notes

The numbers 1 ,2 and 3 define the font size of a markdown cell

First create or go to an existing cell and do 'Esc+ any number of choice' = markdown cell

### Shortcuts to navigate, create, delete and run cells

Press enter to enter a cell, press esc to exit that cell.

Use Arrow keys Up and Down to navigate the cells.

Esc+a = new cell above

Esc+b = new cell above

Esc+dd = deletes the current cell

Ctrl+Enter = runs the current cell and the cursor stays in the same cell

Shift+Enter = runs the current cell and the cursor moves to the below same cell
