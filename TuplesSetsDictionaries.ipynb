### Tuples
Tuples are similar to lists but unlike lists tuples are immutable
Properties:
1. Ordered collections of arbitrary objects: Like strings and lists, tuples are positionally ordered collections of objects (i.e.,they maintain a left-to-right order among their contents); like lists, they can embed any kind of object.
2. Accessed by offset: Like strings and lists, items in a tuple are accessed by offset; they support all the offset-based access operations such as indexing and slicing.
3. Of the category “immutable sequence”: Like strings and lists, tuples are sequences; they support many of the same operations.
However, like strings, tuples are immutable; they don’t support any of the in-place change operations applied to lists.
4. Fixed-length, heterogeneous, and arbitrarily nestable: Because tuples are immutable, you cannot change the size of a tuple without making
a copy. On the other hand, tuples can hold any type of object, including other compound objects (e.g., lists, dictionaries, other tuples), and so support arbitrary nesting.

### Why tuples when we have lists?
1. Tuples were created to use them as a simple association of objects whereas lists can be used as datastructures that change over time (dynamic)
2. Tuples are immutable and hence can be used in places which requires integrity i.e, tuples cannot be changed anywhere during the program whereas lists can
3. Tuples are the only options for storing keys in a dictionary which requires an immutable sequence

### Dictionaries
Dictionaries are unordered collection of objects
1. Unordered collections of arbitrary objects:Unlike in a list, items stored in a dictionary aren’t kept in any particular order; in fact, Python pseudo-randomizes their left-to-right order to provide quick lookup.Keys provide the symbolic (not physical) locations of items in a dictionary
2. Accessed by key, not offset position: Dictionaries are sometimes called associative arrays or hashes (especially by users of other scripting languages). They associate a set of values with keys, so you can fetch an item out of a dictionary using the key under which you originally storedit. You use the same indexing operation to get components in a dictionary as you do in a list, but the index takes the form of a key, not a relative offset.
3. Variable-length, heterogeneous, and arbitrarily nestable: Like lists, dictionaries can grow and shrink in place (without new copies being made), they can contain objects of any type, and they support nesting to any depth (they can contain lists, other dictionaries, and so on). Each key can have just one associated value, but that value can be a collection of multiple objects if needed, and a given value can be stored under any number of keys.
4. Of the category “mutable mapping” : You can change dictionaries in place by assigning to indexes (they are mutable), but they don’t support the sequence operations that work on strings and lists.
Because dictionaries are unordered collections, operations that depend on a fixed positional order (e.g., concatenation, slicing) don’t make sense. Instead, dictionaries are the only built-in, core type representatives of the mapping category—objects that map keys to values. Other mappings in Python are created by imported modules.
