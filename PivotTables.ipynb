```python
import pandas as pd
```

```python
df = pd.read_csv('weather.csv')
df
```

```python
# Pivot helps in transposing rows to columns and vis-a-vis
df.pivot(index='date',columns='city') # to create dates as rows and other data are grouped based on city
```

```python
# to create dates as rows and choose only humidity as values
df.pivot(index='date',columns='city',values='humidity')
```

```python
# or use temperature as values
df.pivot(index='date',columns='city',values='temperature')
```

```python
# make the city as rows
df.pivot(index='city',columns='date',values='temperature')
```

### Pivot tables are used to summarize and aggregate data inside dataframe

```python
df_pivotTable=pd.DataFrame({'date':['5/1/2017','5/1/2017','5/2/2017','5/2/2017',
                     '5/1/2017','5/1/2017','5/2/2017','5/2/2017'],
             'city':['new york','new york','new york','new york',
                    'mumbai','mumbai','mumbai','mumbai'],
             'temperature':[65,61,70,72,75,78,82,82],
             'humidity':[56,54,60,62,80,83,85,26]})
df_pivotTable
```

```python
# How to create a pivot table with average temperature across the cities for both dates
df.pivot_table(index='city',columns='date')
```

### How to use grouper function

```python
df_grouper = pd.DataFrame({'date':['5/1/2017','5/2/2017','5/3/2017',
                                   '12/1/2017','12/2/2017','12/3/2017'],
                          'city':['new york','new york','new york',
                                  'new york','new york','new york'],
                          'temperature':[65,61,70,30,28,25],
                          'humidity':[56,54,60,50,52,51]})
df_grouper
```

```python
# convert the date column to date time object instead of string object
df_grouper['date']=pd.to_datetime(df_grouper['date']) # to convert the date object to date time
type(df_grouper['date'][0]) # to check if the object has been converted
```

```python
df_grouper.pivot_table(index=pd.Grouper(freq='ME',key='date'),columns='city')
```
