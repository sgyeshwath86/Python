### Compound statements : if statement
Compund statements are statements which embed other statements.
If statement is one such example and it may embed other if statements or if-else statements or any other statements
The Python if statement is typical of if statements in most procedural languages. 
It takes the form of an if test, followed by one or more optional elif (“else if”) tests and
a final optional else block.
The tests and the else part each have an associated block of nested statements, indented under a header line.
When the if statement runs, Python executes the block of code associated with the first test that evaluates to true, or the
else block if all tests prove false. The general form of an if statement looks like this

```python
# if-elif-else statements syntax
if test1:        # if test
    statements1  # Associated if block
elif test2:      # Optional elif test
    statements2
else:           # Optional else
    statements3
```

```python
# Example of finding out odd/even number
usrNum = int(input('Please enter a natural number:')) #How to collect user input and perform type casting to convert string input to number
if usrNum%2==0:  # if the test evaluates to True the associated statements get executed
    print(f'Entered number {usrNum} is even')
else: # if the  test above evaluates to false the else statements gets executed
    print(f'Entered number {usrNum} is odd')
```

### Loops

#### for loop

1. The for loop is a generic iterator in Python: it can step through the items in any ordered
sequence or other iterable object.
2. The for statement works on strings, lists, tuples, and other built-in iterables 

```python
# Syntax:
# for iterable:
#     statement1
#     Statemen2
```

```python
# How for loop iterates
nums = [1,2,3,4,5]
for num in nums:
    print(num)
```

```python
# use of break statement in for
nums = [1,2,3,4,5]
for num in nums:
    if num==3:  # Condition to test
        print('Found!!')
        break  #breaks out of the for loop
    print(num) #does not print the number when the if evaluates to True
```

```python
# use of break statement in for
nums = [1,2,3,4,5,3,3,3]
for num in nums:
    if num==3:  # Condition to test
        print('Found!!')
        break  #breaks out of the for loop
    print(num) #does not print the number when the if evaluates to True
```

#### while loop

1. Python’s while statement is the most general iteration construct in the language.
2. Insimple terms, it repeatedly executes a block of (normally indented) statements as long
as a test at the top keeps evaluating to a true value. It is called a “loop” because control
keeps looping back to the start of the statement until the test becomes false.
3. When the test becomes false, control passes to the statement that follows the while block.
The net effect is that the loop’s body is executed repeatedly while the test at the top is true.
If the test is false to begin with, the body never runs and the while statement is skipped.
Syntax:

```python
while test:     # loop test
    statements  # loop body
else:           # optional else
    statements  # runs if the loop didn't exit with break
```

```python
# Running while loop using a counter
x=0  # remember to initiate the counter
while x<10:
    print(x)
    x+=1 # remember to increment the counter, if not the loop will run infinitely
```

```python
# using break in while
x=0
while x<10:
    if x==5:
        break
    print(x)
    x+=1 #x=x+1
```
