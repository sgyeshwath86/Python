```python
import pandas as pd
```

```python
# create a dataframe of temperature across cities of usa
df1 = pd.DataFrame({'city':['NewYork','Chicago','Orlando'],
                   'temperature':[21,14,35]})
df1
```

```python
# create a dataframe of humidity across cities of usa
df2 = pd.DataFrame({'city':['Chicago','NewYork','Orlando'],
                   'humidity':[65,68,75]})
df2
```

```python
# merge the df's using on column method
df_merge = pd.merge(df1,df2,on='city')
df_merge
```

```python
# What if the 2 df's have different cities
```

```python
df3 = pd.DataFrame({'city':['NewYork','Chicago','Orlando','Baltimore'],
                   'temperature':[21,14,35,32]})
df3
df4 = pd.DataFrame({'city':['Chicago','NewYork','Sanfrancisco'],
                   'humidity':[65,68,72]})
df4
```

```python
df_merge_inner = pd.merge(df3,df4,on='city') # default method for how is inner which is basically intersection of the df's
df_merge_inner
# Notice that only the rows having cities in common have been merged, Orlando, Sanfrancisco and Baltimore are left out
```

```python
df_merge_outer = pd.merge(df3,df4,on='city',how='outer') # outer method is basically union of the df's
df_merge_outer
```

```python
df_merge_left = pd.merge(df3,df4,on='city',how='left') # left method takes all the elements in the left df (first in the order). For entries missing in any df NaN is entered
df_merge_left
```

```python
df_merge_right = pd.merge(df3,df4,on='city',how='right') # left method takes all the elements in the right df (second in the order). For entries missing in any df NaN is entered
df_merge_right
```

```python
#  How to check from which dataframe the entries were merged, use indicator=True
df_merge_outer = pd.merge(df3,df4,on='city',how='outer',indicator=True) # outer method is basically union of the df's
df_merge_outer
```

```python
# How to use suffixes when the dataframes have common column names
df5 = pd.DataFrame({'city':['new york','chicago','orlando','baltimore'],
                   'temperature':[21,14,35,38],
                   'humidity':[65,68,71,75]})
df5
```

```python
df6 = pd.DataFrame({'city':['chicago','new york','san diego'],
                   'temperature':[21,14,35],
                   'humidity':[65,68,71]})
df6
```

```python
# When merged, notice that an x is added by default to the column names
df_merge_suffix = pd.merge(df5,df6,on='city')
df_merge_suffix
```

```python
# add suffixes to identify
df_merge_suffix = pd.merge(df5,df6,on='city',suffixes=['_left','_right'])
df_merge_suffix
```
