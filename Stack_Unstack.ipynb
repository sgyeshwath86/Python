### Stack and unstack is another method of reshaping or transforming data

```python
import pandas as pd
```

```python
df = pd.read_excel('stocks_stackUnstack.xlsx') #if we don't provide the header argument the data looks messy as below
df
```

```python
# use the header argument
df = pd.read_excel('stocks_stackUnstack.xlsx',header=[0,1]) # looking at the csv file,in the header we have 2 rows hence providing 0,1
df
```

```python
df_Stacked = df.stack(level=0) #stack the df at a chosen level, default level is -1
df_Stacked
```
