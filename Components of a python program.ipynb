### Expressions and statements

```python
# Expression is a combination of values, variables, operators and function calls
name='Yeshwanth'
designation = 'CAE analyst'
company='TCS'
year = 2021
print(f'MY name is {name} and I work as a {designation} at {company} since {year}')
```

```python
print('My name is:',name,'My designation is',designation)
```

```python
#Statements are a set of instructions for python to execute:
# Assignment statements, Compound assignment statements, Conditional statements, Looping
#There is a subtle difference b/w expressions and statements and most of the time they are used mutually
import random
weather = ['sunny','rainy','cold']
choice = random.choice(weather)
if choice == 'sunny':



    
    print(f"It's {choice}, so shorts and T-shirt")
elif choice == 'rainy':
    print(f"It's {choice}, so grab a rain coat")
elif choice == 'cold':
    print(f"It's {choice}, so stay put at home!!")
```
