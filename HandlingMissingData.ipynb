```python
import pandas as pd
```

```python
# read the csv file, convert the day column to pandas TimeStamp datatype instead of default string
df = pd.read_csv('weather_data_missingValues.csv',parse_dates=['day'])
# print(type(df.day[0]))
df.set_index('day',inplace=True) # set day column as index
df
```

```python
# How to handle NaN values using fillna, dictionary to choose the columns and values to go with
df_copy1 = df.fillna({'temperature':0,
                     'windspeed':0,
                     'event':'no event'})
df_copy1
```

```python
# Instead of filling NaN with 0, use fill method such as forward fill to fill the value from the previous cell of the column
df_copy2 = df.fillna(method='ffill')
df_copy2
```

```python
# Use the bfill method to fill the NaN values from the next cell
df_copy3 = df.bfill()
df_copy3
```

```python
# use axis=columns to copy values row wise from one cell to another
df_copy4 = df.bfill(axis="columns")
df_copy4
```

```python
df
```

```python
# use limit argument to limit the no of times the value can be copied from cell to cell
df_copy5 = df.ffill(limit=1)
df_copy5
```

```python
# how to interpolate values between cells and fill nan
# df.infer_objects(copy=False)
df_copy6 = df.interpolate()
df_copy6
```

```python
# how to interpolate using Time method
df_copy7 = df.interpolate(method="time")
df_copy7
```

```python
df_copy8 = df.dropna(how='all')
df_copy8
```

```python
df_copy9 = df.dropna(thresh=2)
df_copy9
```

```python
# How to create missing date lines
#  Lets add 01-01-2017 to 01-11-2017 days to the existing df
dt_range = pd.date_range('01-01-2017','01-11-2017')
idx = pd.DatetimeIndex(dt_range)
df_10 = df.reindex(idx)
df_10
```
