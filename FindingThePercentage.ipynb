```python
# Finding the percentage
# The provided code stub will read in a dictionary containing key/value pairs of name:[marks] for a list of students.
# Print the average of the marks array for the student name provided, showing 2 places after the decimal.
'''The first line contains the integer , the number of students' records.
The next  lines contain the names and marks obtained by a student, each value separated by a space.
The final line contains query_name, the name of a student to query.'''
#Solution starts....
n = int(input()) # collect the no of students
student_marks = {} #create an empty dictionary
for _ in range(n):   # looping using '_' to collect user entries for n number of times
    name,*line = input().split()  # unpacking to store name and marks (seperated by spaces for ex: Krishna 67 68 69)
    scores = list(map(float,line)) #create a list of the scores entered using map builtin function
    student_marks[name] = scores # add key-value pairs ('name',[marks]) to the dictionary student_marks
query_name = input() #user input for the student name to be queried for average marks
avg_score = sum(student_marks[query_name])/len(student_marks[query_name]) #calculate the average marks, average =(sum of marks)/number of subjects. Here the number of subjects can be calculated using
# the len() function. 
print(f'{avg_score:.2f}') #print the average score to 2 decimal places using f-strings
# Sample Input
#  3
#  Krishna 67 68 69
#  Arjun 70 98 63
#  Malika 52 56 60
#  Malika
#  Sample output
#  56.00
```
