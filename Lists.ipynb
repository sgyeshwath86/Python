## Lists

1. Lists are ordered collection object type and can contain any sort of object: numbers, strings and even other lists
2. Lists are mutable objects which means they can be changed in place

### Properties of lists:
1. Ordered collection: Maintains left to right positional ordering (i.e, they are a sequence)
2. Accessed by offset: Similar to strings, lists can be indexed or sliced
3. Variable length, heterogenrous and nestable: Lists can grow and shrink in place(variable length),they contain any sort of object( heterogeneous) and list of list of lists can be created (nesting)
4. Mutable sequence: supports all sequence operations such as indexing, slicing and concatenation. Lists also support in-place deletion and addition, which changes the lists (mutable)

```python
dir(list)
```

```python
#Methods to grow list
# L3 = ['Bob',40.0,['dev', 'mgr']] 
print(L3)
# L3.append('JLR')
# print('Add JLR to L3',L3) 
# L2 = [123,'abc',1.23,{}]
# print(L2)
# L2.extend([5,6,7]) 
# print('Add [5,6,7] to L2',L2)
L3.insert(2,'single')
print('Add marital status to L3',L3)
```

```python
# Methods to search
print('Index of Bob is',L3.index('Bob'))
print("How many Bob's are in L3?",L3.count('Bob'))
```

```python
# Methods to sort or reverse
L5 = [2,6,1,5,8,100,58]
# print('Before sorting',L5)
# L5.sort() #sorts the list inplace and returns none
# L5
sorted(L5) #sorts the list and returns the sorted list
# print('After sort',L5)
```

```python
# Methods to delete or remove items in a list
# del List[i] i is the index of the list item
print('Before deletion',L5)
del L5[-1]
print('After deletion',L5)
```

```python
print('Before deletion',L5)
del L5[:2]
print('After deletion',L5)
```

```python
print('Before deletion',L5)
L5[0:2]=[]
print('After deletion',L5)
```

```python
# Index or slice assignment. Example to show mutable property of lists
print('L5 before index assignment',L5)
L5[0:2]=[4,5]
print('L5 after index assignment',L5)
```
