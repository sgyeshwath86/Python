```python
# How to read a file using open
f = open('Test.txt')
f.close()
```

```python
print(f.name,f.mode)
```

```python
with open('Test.txt','r') as f: # aliasing 
    f_contents = f.readline()
    print(f_contents,end='')
    f_contents = f.readline()
    print(f_contents)
```

```python
print(f.closed)
```

```python
# How to read the lines of a file one by one using for loop
with open('Test.txt') as f:
    for line in f:
        print(line,end='')
```

```python
# How to read first n sized chunks of a file
with open('Test.txt','r') as f:
    f_contents = f.read(5)
    print(f_contents)
    
```

```python
# How to read contents of a file using while in chunks of a size
with open('Test.txt','r') as f:
    size_to_read=5
    f_contents = f.read(size_to_read)
    while len(f_contents)>0:
        print(f_contents,end='*')
        f_contents=f.read(size_to_read)
```

```python
# How to find out the cursor position
with open('Test.txt','r') as f:
    size_to_read=5
    f_contents = f.read(size_to_read)
    print(f.tell()) #points out the cursor position
```

```python
# How to choose the cursor position to start from the beginning of the line
with open('Test.txt','r') as f:
    size_to_read=10
    f_contents = f.read(size_to_read)
    print(f_contents)
    f.seek(5) # places the cursor at the beginning of the file
    f_contents = f.read(size_to_read)
    print(f_contents)
```

```python
# How to create a new file using write
with open('test2.txt','w') as f:
    f.write('Test')
    f.seek(2)
    f.write('R')
```

```python
# How to write the contents of an existing file to a new file
with open('Test.txt','r') as rf:
    with open('Test_copy.txt','w') as wf:
        for line in rf:
            wf.write(line)
```
