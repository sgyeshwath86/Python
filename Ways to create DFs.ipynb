```python
import pandas as pd
```

```python
# using dictionaries
weather_data = {
    'day':['1/1/2017','1/2/2017','1/3/2017','1/4/2017','1/5/2017','1/6/2017'],
    'temperature':[32,35,28,24,32,31],
    'windspeed':[6,7,2,7,4,2],
    'event':['Rainy','Sunny','Snow','Snow','Rain','Sunny']
    
}
df_weather = pd.DataFrame(weather_data)
df_weather
```

```python
# how to fetch head and tails of a df
df_weather.head(5)
df_weather.tail(1)
```

```python
# the shape of the df can be obtained from df.shape which gives a tuple
rows,cols=df_weather.shape
print(f'The rows and columns of the df are {rows} and {cols}')
```

```python
# indexing in pandas
print(df_weather[2:5])
print(df_weather[:3])
```

```python
# how to fetch the column names from a df
df_weather.columns
```

```python
# print the individual chosen column
print(df_weather['day']) #method 1
print(df_weather.day) #method 2
```

```python
# how to check the type of data in a df
type(df_weather['day']) # all columns in a df are called as pandas series
```

```python
# how to output selected columns (more than 1) from a dataframe
df_weather[['day','temperature']]
```

```python
# how to perform some basic operations on the columns of a df
print('Maximum temperature',df_weather['temperature'].max())  #to get the maximum value from the column
print('Average temperature',df_weather.temperature.mean()) # to get the mean or average value
print('Standard deviation',df_weather['temperature'].std())
df_weather.describe() # gives the statistical info for the columns containing numerical data
```

```python
# how to perform conditional operations on the columns of interest
df_weather[df_weather['temperature']>=32]  #provide me days where the temperature was >=32
df_weather[df_weather['temperature']==df_weather['temperature'].max()] #give me the days where the temperature was == to the max temperature
df_weather['day'][df_weather['temperature']==df_weather['temperature'].max()] # give me only the day not the entire row where temperature == max
df_weather[['temperature','day']][df_weather['temperature']==df_weather['temperature'].max()] # give me the day and the temperature value in the df where the temperature == max
```

```python
# how to set index for a df other than the default index
# df_weather.set_index('day',inplace=True)
df_weather
```

```python
# Now the date can be used as index and corresponding rows can be fetched using the loc function
df_weather.loc['1/5/2017']
```

```python
# If you want the original index, use reset index along with inplace=True
df_weather.reset_index(inplace=True)
df_weather
```

```python
# set event as index and fetch rows where it is snow
df_weather.set_index('event',inplace=True)
df_weather.loc['Snow']
```
