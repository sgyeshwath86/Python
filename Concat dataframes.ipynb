```python
import pandas as pd
```

```python
# create a dataframe of cities and weather data
india_weather = pd.DataFrame({'city':['Mumbai','Delhi','Bangalore'],
                  'temperature':[32,45,30],
                  'humidity':[80,60,78]})
india_weather
```

```python
usa_weather = pd.DataFrame({'city':['Newyork','Chicago','Orlando'],
                  'temperature':[21,14,35],
                  'humidity':[68,65,75]})
usa_weather
```

```python
# How to concat 2 dataframes
df = pd.concat([india_weather,usa_weather]) #this creates a df with individual indices taken from the dataframes
df
```

```python
df = pd.concat([india_weather,usa_weather],ignore_index=True) #this creates a df with continous indices
df
```

```python
df = pd.concat([india_weather,usa_weather],keys=['india','usa'],ignore_index=False) #this creates a df with additional indices of choice using keys argument
df
```

```python
# the advantage of using keys is we can sort out the data based on keys as below using df.loc
df.loc['india']
```

```python
# Create a dataframe of temperature data
temperature_df = pd.DataFrame({'city':['mumbai','delhi','bangalore'],
                              'temperature':[32,45,30]},index=[0,1,2])
temperature_df
```

```python
# Create a dataframe of windspeed data
windspeed_df = pd.DataFrame({'city':['delhi','mumbai'],
                              'windspeed':[12,15]},index=[1,0])
windspeed_df
```

```python
# How to combine the temperature and windspeed df's to a single df so that the columns appear side by side
# if we use the pd.concat it won't work, the common rows will not be combined
df_concat = pd.concat([temperature_df,windspeed_df])
df_concat
```

```python
# if we use axis = 1
df_concat_1 = pd.concat([temperature_df,windspeed_df],axis=1)
df_concat_1
# The combined df does not appear correct since the temperature and wind speed of delhi and mumbai are mismatched
# Use index argument in the individual dataframes to correct this
```

```python
# How to combine a series to a dataframe
# Lets create a series for events of the cities
event_Series = pd.Series(['Humid','Dry','Rain'],name="event")
event_Series
```

```python
df_temp=pd.concat([temperature_df,event_Series],axis=1)
df_temp
```
