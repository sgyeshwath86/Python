### Crosstab functionality in pandas helps us in creating contingency tables which can be used for statistical analysis, especially for multivariate frequency distribution of variables

```python
import pandas as pd
```

```python
df = pd.read_excel('Crosstab.xlsx')
df
```

```python
# Create a table showing the handedness of people across nations
pd.crosstab(df.Nationality,df.Handedness)
```
