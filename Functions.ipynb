### Functions
1. A function is a device that groups a set of statements so they can be run more than once in a program—a packaged procedure invoked by name.
2. Functions are also the most basic program structure Python provides for maximizing code reuse, and lead us to the larger notions of program design.

```python
# Let's explore return
def sum(a,b):    # using parameters/arguments a and b
    output=a+b
    return output  # returning the sum as an output
```

```python
#How to pass arguments
def greeting(msg):
    return '{} World.'.format(msg)
```

```python
print(greeting('Hi')) # Calling a function without the required arguments gives an error
```

```python
print(greeting('Hi'))
```

```python
# Required and keyword positional arguments
def greeting_name(msg,name='User'):
    return '{},{}'.format(msg,name)
```

```python
greeting_name() # Still throws an error when the required positional argument is empty
```

```python
greeting_name('Hello') # Runs without any error even when the keyword positional argument is empty
```

```python
greeting_name('Hello','Yesh') # When the keyword positional argument is given, the default argument is overwritten
```

```python
# How to use *args and *kwargs
# postional aruments (*args) always preceds keyword arguments(**kwargs)
def student_info(*args,**kwargs):  #*pos **keyword
    print(args,type(args))
    print(kwargs,type(kwargs))
```

```python
# Lets add some positional and keyword args to the defined function above
student_info('Math','Art',name='John',age=25)
# student_info('Math','Art',name='dummy','Science')
```

```python
def cars(*vehCode,**vehName):
    print(vehCode)
    print(vehName)
```

```python
vehCode = ['RR_Velar','Discovery','RR_Evoque','Jaguar']
vehName = {'code1':'L481','code2':'L482','code3':'L381','code4':'X900'}
```

```python
cars(vehCode,vehName) # unpacking is not working as it is supposed
```

```python
cars(*vehCode,**vehName) # Assigns the list(vehCode) to args and dictionary(vehName) to kwargs
```

```python
# Example to show the above concepts
# Number of days per month. First value placeholder for indexing purposes.
month_days = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]


def is_leap(year):
    """Return True for leap years, False for non-leap years."""  # This is the doc string of the function

    return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)


def days_in_month(year, month):
    """Return number of days in that month in that year."""

    if not 1 <= month <= 12:
        return 'Invalid Month'

    if month == 2 and is_leap(year):
        return 29

    return month_days[month]
```
